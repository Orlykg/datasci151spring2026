```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
pres_county = pd.read_csv("4-US_Election_2020/president_county.csv")
pres_cand = pd.read_csv("4-US_Election_2020/president_county_candidate.csv")

senate_county = pd.read_csv("4-US_Election_2020/senate_county.csv")
senate_cand = pd.read_csv("4-US_Election_2020/senate_county_candidate.csv")

gov_county = pd.read_csv("4-US_Election_2020/governors_county.csv")
gov_cand = pd.read_csv("4-US_Election_2020/governors_county_candidate.csv")
```

```python
pres_cand = pres_cand.query("party == 'DEM' or party == 'REP'")

# 2. Filter Senate candidates
senate_cand = senate_cand.query("party == 'DEM' or party == 'REP'")

# 3. Filter Governor candidates
gov_cand = gov_cand.query("party == 'DEM' or party == 'REP'")
```

```python
pres_merged = pd.merge(pres_cand, pres_county[['state', 'county', 'total_votes']], on=['state', 'county'])

# 2. Merge Senate files (Both use 'total_votes')
sen_merged = pd.merge(senate_cand, senate_county[['state', 'county', 'total_votes']], on=['state', 'county'])

# 3. Merge Governor files (FIXED)
# We use 'total_votes' here because that's what the governor_county file calls it!
gov_merged = pd.merge(gov_cand, gov_county[['state', 'county', 'total_votes']], on=['state', 'county'])
```

```python
my_bins = [0, 50000, 1000000, np.inf]

# 2. We give each 'bucket' a name
my_labels = ['Rural', 'Suburban', 'Urban']

# 3. We use pd.cut to sort the counties into these buckets
# Remember: we use 'total_votes_y' because that is the total county size!
pres_merged['size_category'] = pd.cut(pres_merged['total_votes_y'], bins=my_bins, labels=my_labels)

# 4. Calculate the percentage (Candidate votes / County total)
pres_merged['percentage'] = (pres_merged['total_votes_x'] / pres_merged['total_votes_y']) * 100
```

```python
pres_result = pres_merged.groupby(['size_category', 'party'])['percentage'].mean()

print("Presidential Election Average Results")
print(pres_result)
```
